Hago lo mismo para el resto de Data Frames y como son todos iguales, voy a crear una función para solo tener que lanzarla en resto de data frames



```python
#Creo la función para limpiar los datos
def limpieza_datos(df):
    #crearé una copia para no cargarme el original
    df_limpio = df.copy()
    #corrige la fecha 
    df_limpio["DATA LANÇAMENTO"] = pd.to_datetime(df_limpio["DATA LANÇAMENTO"], format="%d/%m/%Y")
    
    # Convertir columnas monetarias y porcentuales
    columnas_a_convertir = ["VALOR PREVISTO ATUALIZADO", "PERCENTUAL REALIZADO", 
                            "VALOR LANÇADO", "VALOR REALIZADO"]
    
    for columna in columnas_a_convertir:
        if columna in df_limpio.columns:
            # Reemplazar comas por puntos
            df_limpio[columna] = df_limpio[columna].astype(str).str.replace(",", ".")
            # Convertir a float
            df_limpio[columna] = df_limpio[columna].astype(float)
    
    return df_limpio
```

```python
#También creo una función para comprobar duplicados
def comprueba_dup(df):
    duplicados_df = df.duplicated().sum()
    return  print(duplicados_df)
```

```python
#Importo 2014 y le aplico la función 
datos_2014 = pd.read_csv("datos-2014.csv", sep = ";", encoding = "latin-1")
datos_2014.head(3)
```

```python
datos_2014 = limpieza_datos(datos_2014)
```

```python
datos_2014.info()
```

```python
comprueba_dup(datos_2014)
```

Vemos que existen valores nulos en la fecha de lanzamiento, haremos un pequeño análisis para ver cuales son y si los podemos rellenar de alguna forma

```python
filtro_nulos = datos_brasil[datos_brasil["DATA LANÇAMENTO"].isnull()]
filtro_nulos.head(5)
```

```python
filtro_nulos_año = filtro_nulos.groupby(filtro_nulos["ANO EXERCÍCIO"]).count("")
filtro_nulos_año
```

```python
#Vemos que empiezan a partir de 2016, contaré el valor de los nulos sobre el total
medida = datos_brasil.shape
porcentaje_nulos = datos_brasil["DATA LANÇAMENTO"].isnull().sum() / medida[0] * 100
porcentaje_nulos = float(porcentaje_nulos)
porcentaje_nulos
```

el porcentaje de nulos es del 0.05% sobre el total de registro y está repartido en varios años por lo que decido quitar los valores nulos ya que no podemos obtenerlos de otras columnas y no son significativos para nuestro análisis ni parece que vengan de un solo año.

```python
brasil_datos_limpio = datos_brasil.dropna(subset=["DATA LANÇAMENTO"])
brasil_datos_limpio.info()
```

```python
#Ahora que no hay nulos, vamos a ver la naturaleza de las columnas y si son coherentes
brasil_datos_limpio.describe().T
```

```python
#Convertiré las columnas de codigo a objeto ya que tiene más sentido puesto que los códigos van relacionados con los nombres
brasil_datos_limpio["CÓDIGO ÓRGÃO SUPERIOR"] = brasil_datos_limpio["CÓDIGO ÓRGÃO SUPERIOR"].astype("object")
brasil_datos_limpio["CÓDIGO ÓRGÃO"] = brasil_datos_limpio["CÓDIGO ÓRGÃO"].astype("object")
brasil_datos_limpio["CÓDIGO UNIDADE GESTORA"] = brasil_datos_limpio["CÓDIGO UNIDADE GESTORA"].astype("object")
```

```python
brasil_datos_limpio.describe(include=object).T
```

Las columnas de codigo orgao y nombre de orgao deberían tener los mismos valores vemos que hay algunas organizaciones con varios códigos, lo mismo pasa con la unidad gestora. Como lo que nos importa son los nombres, eliminare las columnas de códigos para evitar confusiónes 

```python
brasil_datos_limpio = brasil_datos_limpio.drop(columns=['CÓDIGO ÓRGÃO SUPERIOR', 'CÓDIGO ÓRGÃO', 'CÓDIGO UNIDADE GESTORA'])
brasil_datos_limpio.info()
```
